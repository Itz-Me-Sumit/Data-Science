```python
import tensorflow as tf
from tensorflow import keras
from keras import layers , Sequential
```

```python
ZFNet = Sequential([

    layers.Input(shape=(244,244,3)),

    layers.Conv2D(
        96,
        kernel_size=(7,7),
        strides=2,
        activation="relu",
    ),

    # LRN
    layers.Lambda(
        lambda x: tf.nn.local_response_normalization(x)
    ),

    layers.MaxPooling2D(
        pool_size=(3,3),
        strides=2
    ),

    layers.Conv2D(
        256,
        kernel_size=(3,3),
        activation="relu"
    ),

    # LRN
    layers.Lambda(
        lambda x: tf.nn.local_response_normalization(x)
    ),

    layers.MaxPooling2D(
        pool_size=(3,3),
        strides=2
    ),

    layers.Conv2D(
        384,
        kernel_size=(3,3),
        padding="same",
        activation="relu"
    ),

    layers.Conv2D(
        384,
        kernel_size=(3,3),
        padding="same",
        activation="relu"
    ),

    layers.Conv2D(
        256,
        kernel_size=(3,3),
        padding="same",
        activation="relu"
    ),

    layers.MaxPooling2D(
        pool_size=(3,3),
        strides=2
    ),

    layers.Flatten(),

    layers.Dense(
        4096,
        activation="relu"
    ),

    layers.Dropout(0.5),

    layers.Dense(
        4096,
        activation="relu"
    ),

    layers.Dropout(0.5),

    layers.Dense(
        1000,
        activation="softmax"
    )
])

ZFNet.summary()
```
