```python
import tensorflow as tf
from tensorflow import keras
from keras.layers import *
```

```python
Flow1_Input = Input(shape = (128,))
Flow1_Dense1 = Dense(128 , activation = "relu")(Flow1_Input)
Flow1_Dense2 = Dense(64 , activation = "relu")(Flow1_Dense1)
Flow1_Dense3 = Dense(32 , activation = "relu")(Flow1_Dense2)
```

```python
Flow2_Input = Input(shape = (32,))
Flow2_Dense1 = Dense(16 , activation = "relu")(Flow2_Input)
Flow2_Dense2 = Dense(8 , activation = "relu")(Flow2_Dense1)
```

```python
Merged = Concatenate()([Flow1_Dense3 , Flow2_Dense2])
```

```python
Flow_Merged = Dense(8 , activation = "relu")(Merged)
Flow_Merged = Dense(1 , activation = "sigmoid")(Flow_Merged)
```

```python
model = Model(
    inputs = [Flow1_Input , Flow2_Input],
    outputs = [Flow_Merged]
)
```

```python
from keras.utils import plot_model
plot_model(
    model,
    show_shapes = True
)
```
