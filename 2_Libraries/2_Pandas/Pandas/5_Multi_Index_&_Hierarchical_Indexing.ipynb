### For multi-Dimensional Data we use Multi-indexing , we use multiple of column's index as once

```python
import numpy as np
import pandas as pd
```

```python
arrays = [
    ['North' , 'North' , 'South' , 'South' , "East"],
    [2024 , 2025 , 2024 , 2025 , 2026]
]
```

```python
index = pd.MultiIndex.from_arrays(arrays , names=("Region","Year"))
```

```python
index
```

```python
df = pd.DataFrame(
    np.random.randint(100 , 500 , size=(5,2)),
    index=index,
    columns=["Products_A","Products_B"]
)
df
```

```python
df.xs('North' , level="Region")
```

```python
df.xs(2024 , level="Year")
```
