````python
# create DF

left = pd.DataFrame({
    'key1':['K0', 'K0', 'K1', 'K2'],
    'key2': ['K0', 'K1', 'K0', 'K1'],
    'A': ['A0', 'A1', 'A2', 'A3'],
    'B':['B0', 'B1', 'B2', 'B3']
})

right = pd.DataFrame({
    'key1':['K0', 'K1','K1','K2'],
    'key2': ['K0', 'K0', 'K0', 'K0'],
    'C': ['C0', 'C1', 'C2', 'C3'],
    'D': ['D0', 'D1', 'D2', 'D3']
})

print("\nThe DataFrame 'left'\n",'-'*30, sep='')
print(left)

print("\nThe DataFrame 'right'\n",'-'*30, sep='')
print(right)
````

````python
merge2 = pd.merge(
    left , right,
    on=["key1","key2"]
)
merge2
````

#### Left Merge

````python
left_merge = pd.merge(
    left , right,
    on=["key1" , "key2"],
    how="left"
)
left_merge
````

#### Right Merge

````python
right_merge = pd.merge(
    left , right,
    on=["key1","key2"],
    how="right"
)
right_merge
````

````python
outer_merge = pd.merge(
    left , right,
    on=["key1" , "key2"],
    how="outer"
)
outer_merge
````

````python
left = pd. DataFrame(
    {
        'A': ['A0', 'A1', 'A2'],
        'B': ['BO', 'B1', 'B2']
    },
    index=['K0', 'K1', 'K2']
)

right = pd.DataFrame(
    {
        'C': ['co', 'C2', 'C3'],
        'D': ['DO', 'D2', 'D3']
    },
    index=['K0', 'K2', 'K3'])

print("\nThe DataFrame 'left'\n", '-'*30, sep='')
print(left)

print("\nThe DataFrame 'right'\n",'-'*30, sep='')
print(right)
````

#### Left Join (Similar as Left Outer Join)

````python
left.join(right)
````

#### Right Join (Similar to Right Outer Join)

````python
right.join(left)
````
