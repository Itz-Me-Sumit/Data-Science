### Introduction to Clustering & A New Approach

Clustering ka main maksad un objects ko ek group mein dalna hai jinki properties ek jaisi hain. 
**Example (Height vs Weight):** Agar ek graph mein X-axis par Weight aur Y-axis par Height ho, toh points ka ek cluster un logo ko represent karega jinka BMI (Body Mass Index) kam hai (yani height zyada par weight kam). Dusra cluster high BMI walo ka ho sakta hai. Ek hi cluster ke andar ke log ek dusre ke zyada "kareeb" (similar) hote hain bajaye dusre cluster ke logo ke.

Ab tak humne **K-Means** padha tha. Ab hum padhenge **Hierarchical Clustering**. Iske do tareeqe hote hain:
1. **Bottom-Up (Agglomerative - HAC):** Isme shuru mein har ek data point ko apna khud ka ek alag cluster maan liya jata hai. Fir jo do clusters sabse paas hote hain, unko jodkar ek bada cluster bana diya jata hai. Ye tab tak chalta hai jab tak sab jud kar ek single bada cluster na ban jayein.
2. **Top-Down (Divisive):** Ye bilkul ulta hai. Shuru mein pura data ek hi bada cluster hota hai, aur fir hum usko tod-tod kar chote clusters banate hain.

Is colab mein hum **HAC (Bottom-Up)** par focus karenge.

### Distance Metrics (Similarity Kaise Naapein?)

Clusters ko jodne ya todne ke liye hume pata hona chahiye ki wo ek dusre ke kitne kareeb hain. Iske liye hum **Metrics (Distance)** ka use karte hain. Ek valid distance metric hone ke liye usme ye 3 properties honi chahiye:
1. **Non-negative:** Doori kabhi negative nahi ho sakti. ($d \ge 0$)
2. **Symmetric:** $A$ se $B$ ki doori wahi hogi jo $B$ se $A$ ki hogi. ($d(A,B) = d(B,A)$)
3. **Triangle Inequality:** Agar aap $A$ se $B$ jate hain, toh seedha rasta hamesha kisi teesre point $C$ se ghoom kar jane wale raste se chota (ya barabar) hoga. ($d(A,B) \le d(A,C) + d(C,B)$)

#### Popular Distance Formulas (The Math):
Yahan $x^{(i)}$ aur $x^{(j)}$ do alag-alag data points (vectors) hain.

1. **Euclidean Distance:** (Seedhi line ki doori, Pythagoras theorem par based)
   $$d(x^{(i)}, x^{(j)}) = \sqrt{\sum_{k=1}^{n} \left(x^{(i)}_k - x^{(j)}_k\right)^2}$$

2. **Manhattan Distance:** (Grid ya blocks par chalne wali doori, jaise shahar ki sadkon par chalna)
   $$d(x^{(i)}, x^{(j)}) = \sum_{k=1}^{n} \left|x^{(i)}_k - x^{(j)}_k\right|$$

3. **Cosine Distance:** (Do points ke beech ka angle dekhta hai. Agar angle 0 hai toh similarity sabse zyada hai)
   $$d(x^{(i)}, x^{(j)}) = 1 - \cos(\theta) = 1 - \frac{x^{(i)} \cdot x^{(j)}}{\|x^{(i)}\| \|x^{(j)}\|}$$

### Linkage (Groups ko Jodne ki Strategy)

Jab har point khud mein ek cluster ban gaya, toh kin do groups ko joda jaye? Iske liye hum **Linkage** ka use karte hain. Ye 4 types ka hota hai:

1. **Single Linkage:** Do groups ke beech ki sabse **choti** doori ko unka distance manta hai. (Nearest neighbors).
   $$L(C_1, C_2) = \min_{x \in C_1, y \in C_2} d(x, y)$$
2. **Complete Linkage:** Do groups ke beech ki sabse **badi** (maximum) doori ko unka distance manta hai. (Furthest elements). Isse clusters tight bante hain.
   $$L(C_1, C_2) = \max_{x \in C_1, y \in C_2} d(x, y)$$
3. **Average Linkage:** Do groups ke saare points ke beech ki saari possible dooriyon ka average nikalta hai.
4. **Ward's Linkage:** Ye distance nahi dekhta, ye Variance (error) dekhta hai. Do groups ko jodne ke baad unke andar ki "Sum of Squared Distances" (galti) kitni kam badhti hai, uspe kaam karta hai. (Ye sabse popular aur accha result dene wala method hai).

### HAC Algorithm in Action (Kaise Kaam Karta Hai?)

**Algorithm ka simple loop:**
1. Pehle saare clusters ke beech ka Distance Matrix calculate karo (matlab sabki sabse doori napo).
2. **While** (jab tak saare points jud kar 1 single cluster na ban jayein):
   * a. Do sabse kareebi clusters ko dhundho (using Linkage).
   * b. Un dono ko merge (jod) kar ek naya cluster bana do.
   * c. Naye cluster aur bache hue baaki clusters ke beech ki doori wapas update karo.

---

### Dendrogram (The Map of Merging)

**Dendrogram** ek ped (tree) jaisa graph hota hai jo batata hai ki kaunsa point kiske sath, kis level par (kitni doori par) juda hai. X-axis par data points hote hain, aur Y-axis par wo doori (distance) hoti hai jispar wo jude hain. Isko dekh kar hum decide kar sakte hain ki hume kitne clusters chahiye (by drawing a horizontal line to cut the branches).

### Toy Dataset & Python Implementation

Chaliye is puri theroey ko code mein dekhte hain.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import normalize
import scipy.cluster.hierarchy as shc

# Step 1: Toy Data Banana
X = np.array([(8,3),(5,3),(6,4),(1,6),(2,8)])

# Step 2: Data Normalize Karna (Scale down karna)
scaled_X = normalize(X)

# Scatter plot se data points dekhna
plt.scatter(X[:,0], X[:,1])
plt.show()

# Step 3: Dendrogram Draw Karna
plt.figure(figsize=(8, 8))
plt.title("Dendrogram")

# shc.linkage model banata hai aur shc.dendrogram usko draw karta hai
# Humne 'ward' linkage use kiya hai
dend = shc.dendrogram(shc.linkage(scaled_X, method='ward'))
plt.show()
```

#### Code Logic:
* **`normalize(X)`**: Data ko ek scale par lata hai taki doori calculate karne mein koi dimension haawi (dominate) na ho jaye.
* **`scipy.cluster.hierarchy (shc)`**: Scikit-learn ke alawa Scipy library Dendrograms draw karne ke liye best hai.
* **`shc.linkage(..., method='ward')`**: Ye function exactly wahi Ward's linkage math perform kar raha hai jo humne upar discuss ki. Ye saare points ko merge karke ek sequence banata hai, jise dendrogram function automatically ek ped ki tarah draw kar deta hai.
