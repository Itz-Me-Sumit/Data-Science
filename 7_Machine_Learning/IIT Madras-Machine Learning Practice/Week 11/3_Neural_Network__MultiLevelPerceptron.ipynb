### Introduction to Neural Networks: Multilayer Perceptron (MLP)

**Neural Network** insaani dimag se inspire hokar banaya gaya ek mathematical model hai. Insaani dimag mein 'Neurons' hote hain jo information pass karte hain. Usi tarah Artificial Neural Network mein 'Nodes' (ya artificial neurons) hote hain. 

**Multilayer Perceptron (MLP)** sabse basic aur classic form hai ek Neural Network ka. Isme normally 3 tarah ki layers hoti hain:
1. **Input Layer:** Jahan se data (jaise image ke pixels) network mein enter karta hai.
2. **Hidden Layer(s):** Ye input layer aur output ke beech ki layers hoti hain. Yahi par asli computation aur 'learning' hoti hai. Isme ek ya ek se zyada layers ho sakti hain.
3. **Output Layer:** Ye final answer (prediction) deti hai.



---

### Key Concepts Explained (The Logic & Math)

#### 1. It is a Supervised Learning Algorithm
**Logic:** Supervised learning ka matlab hai ki "Teacher" model ko sikhata hai. Hum MLP ko data (X) ke sath-sath sahi answers ya labels (y) bhi dete hain. Model apni prediction karta hai, galti (error) ko sahi answer ke sath compare karta hai, aur apne andar ke parameters ko update karta hai taki agli baar galti kam ho. Is galti sudharne ke process ko **Backpropagation** kehte hain.

#### 2. MLP learns a non-linear function approximator
**Logic:** Duniya ki bohot sari problems ek seedhi line (linear function) se solve nahi ki ja sakti. Ek basic linear equation aisi dikhti hai: 
$$y = W \cdot X + b$$
(Jahan $W$ weight hai aur $b$ bias hai). Agar data bohot complex hai (jaise dog aur cat ki image me difference pehchanna), toh ek seedhi line unko alag nahi kar sakti. 

Isiliye MLP mein **Activation Functions** (jaise ReLU, Sigmoid) lagaye jate hain jo is mathematical equation ko "tedha-medha" (non-linear) bana dete hain. **Function Approximator** ka matlab hai ki MLP ek aisi complex mathematical equation (function) dhundh (approximate kar) leta hai jo diye gaye dataset par bilkul fit baithti hai.

### Sklearn Implementation: Classifier vs Regressor

Scikit-Learn (sklearn) library mein MLP ko do alag-alag modules ke zariye implement kiya jata hai, problem ke type ke hisab se:

### Deep Dive into MLPClassifier Capabilities

#### 1. Multi-class Classification & The Softmax Function
**Logic:** Multi-class ka matlab hai ki aapke paas 2 se zyada classes hain, **par ek sample sirf ek hi class mein ja sakta hai**. (Example: Image ya toh Apple hai, ya Banana, ya Orange. Ek sath teeno nahi ho sakti).

Aise mein Output layer par **Softmax** function lagaya jata hai. 
**The Math Behind Softmax:** Softmax function network se aane wale raw numbers (logits) ko aisi probabilities mein badal deta hai jinka total hamesha 1 (ya 100%) hota hai. 
$$P(y_i) = \frac{e^{z_i}}{\sum_{j=1}^{K} e^{z_j}}$$
* $z_i$: Class $i$ ka raw output number.
* $e$: Euler's number (approx 2.718). Exponential lene se negative numbers bhi positive ho jate hain aur bada number aur zyada bada ho jata hai.
* Denominator (Niche ka hissa): Saare classes ke exponential ka sum, jo ensure karta hai ki sari probabilities milakar 1 ban jayein.
*(Agar Apple ki probability 0.8 aayi, toh iska matlab model 80% sure hai ki ye Apple hai).*

#### 2. Multi-label Classification
**Logic:** Multi-label ka matlab hai ek sample ek se zyada classes (labels) mein fit ho sakta hai. 
(Example: Ek movie ek hi waqt par 'Action', 'Comedy', aur 'Sci-Fi' teeno ho sakti hai). 
Is case mein, hum categories ke beech competition nahi karate (yani Softmax use nahi karte). Iski jagah har category apni alag probability (usually Sigmoid function se) nikalati hai, jaise: Action (90%), Comedy (70%), Romance (10%).

### Important Parameters (Model ko Apne Hisab Se Customize Karna)

#### 1. `hidden_layer_sizes` (Layers aur Neurons set karna)
* Ye batata hai ki kitni layers hongi aur har layer mein kitne neurons (nodes) honge.
* Ise ek Tuple `()` mein likha jata hai.
* **Example:** `MLPClassifier(hidden_layer_sizes=(15, 10, 5))`
  *(Matlab: 3 Hidden layers. Pehli mein 15 neurons, dusri mein 10, aur teesri mein 5).*

#### 2. `alpha` (Regularization - Overfitting Rokna)
* Ye L2 penalty (Ridge Regularization) lagata hai taki model data ko ratne (overfit) na lag jaye.
* **Default value:** `0.0001` (Aap ise badha sakte hain agar model overfit kar raha ho).

#### 3. `activation` (Hidden Layers ka Math Function)
Ye function decide karta hai ki node se data pass hoga ya nahi. Options hain:

* **`'relu'` (Default & Best):** $f(x) = \max(0, x)$. Negative ko 0 banata hai, positive ko waisa hi rakhta hai.
* **`'logistic'` (Sigmoid):** $f(x) = \frac{1}{1 + e^{-x}}$. Output ko 0 aur 1 ke beech dabata hai.
* **`'tanh'`:** Output ko -1 aur 1 ke beech dabata hai.
* **`'identity'`:** Jo input aaya, wahi bhej do ($f(x) = x$). Koi maths nahi.

### Weight Optimization (Model Kaise Seekhta Hai?)

Neural Network galti (log-loss) ko kam karne ke liye apne 'weights' ko lagatar adjust karta hai. Is kaam ke liye hum `solver` parameter ka use karte hain. 

#### 1. `solver` (The Optimizer)
* **`'adam'` (Default):** Zyada tar datasets ke liye sabse fast aur best optimization karta hai.
* **`'sgd'`:** Stochastic Gradient Descent. Ye basic aur standard tareeqa hai.
* **`'lbfgs'`:** Ye chote datasets ke liye bohot accha kaam karta hai. (Dhyan rahe: Ye solver minibatches par kaam nahi karta, pura data ek sath dekhta hai).

#### 2. `batch_size` (Kitna data ek baar mein padhe?)
* `'sgd'` aur `'adam'` pura data ek sath dekhne ki jagah usko chote-chote tukdo (batches) mein padhte hain.
* **Default:** `batch_size = min(200, n_samples)` (Yani agar 1000 data points hain, toh ek baar mein 200 padhega. Par agar data hi 150 hai, toh 150 padhega).

```python
# Input layer aur 1st Hidden layer ke beech ke weights dekhne ke liye:
print(MLP_clf.coefs_[0])

# 1st Hidden layer aur 2nd Hidden layer ke beech ke weights dekhne ke liye:
print(MLP_clf.coefs_[1])

### Trained Model ke Biases (Intercepts) Kaise Dekhein?

Jaise humne model ke weights dekhne ke liye `coefs_` ka use kiya tha, waise hi model ke andar ke **Bias** (har node ka apna extra constant addition) ko dekhne ke liye hum `intercepts_` ka use karte hain.

* **`intercepts_`**: Ye bhi ek list hoti hai jiska size `(n_layers - 1)` hota hai.
* Is list ka har ek element (i) agli layer (i+1) ke sabhi nodes ka bias vector hota hai.

**Examples:**
```python
# First hidden layer ke nodes ka bias dekhne ke liye:
print(MLP_clf.intercepts_[0])

# Second hidden layer ke nodes ka bias dekhne ke liye:
print(MLP_clf.intercepts_[1])

### MLPClassifier: Learning Rate & Iteration Parameters

Ye parameters batate hain ki model kitni tezi se (speed) aur kab tak seekhega.

#### 1. `learning_rate` (Sirf `solver='sgd'` ke liye)
Model ke seekhne ki speed (step size) training ke dauran kaise badlegi:
* **`'constant'` (Default):** Speed shuru se aakhiri tak ek jaisi rahegi.
* **`'invscaling'`:** Jaise-jaise time badhega, speed dheere hoti jayegi. (Is calculation ke liye `power_t` parameter use hota hai, jiska default value `0.5` hai).
* **`'adaptive'`:** Jab tak error kam ho raha hai, speed same rahegi. Agar error kam hona ruk jaye, toh ye automatically speed ko chota kar deta hai.

#### 2. `learning_rate_init` (For `'sgd'` & `'adam'`)
* **Default = 0.001.** Ye batata hai ki training ke pehle step mein seekhne ki shuruati (initial) speed kya hogi.

#### 3. `max_iter`
* **Default = 500.** Model zyada se zyada kitni baar (epochs/iterations) pure data ko dekhega aur apne aap ko update karega.

---

### Extra Rules (Kaun parameter kiske sath kaam karta hai?)
* **`power_t`:** Sirf tab kaam aayega jab aapka `solver='sgd'` ho aur `learning_rate='invscaling'` ho.
* **`shuffle`:** Har iteration mein data ko mix (shuffle) karne ke liye taki model ratein na. Ye sirf `'sgd'` aur `'adam'` mein kaam karta hai.
* **`momentum`:** Model ko kisi ghalat jagah atakne se bachane aur tezi dene ke liye. Ye bhi sirf `solver='sgd'` ke sath use hota hai.

### MLPRegressor (Continuous Values Predict Karna)

Jahan `MLPClassifier` classes (Categories) batata hai, wahi `MLPRegressor` lagatar badalne wale numbers (Continuous values, jaise: Ghar ki price, temperature) predict karne ke liye use hota hai.

#### Asli Farak Kya Hai? (The Logic)

1. **No Output Activation:** Classifier ke output layer mein humne dekha tha ki Softmax ya Sigmoid lagate hain taki answer probability (0 se 1) ke beech aaye. Par Regressor ke output layer mein **koi activation function nahi hota** ($f(x) = x$). Kyun? Kyunki hume exact number chahiye, kisi range mein daba hua answer nahi.

2. **Loss Function (Galti napne ka tareeqa):**
   Classifier "Cross-Entropy" use karta hai. Par Regressor **Square Error (MSE - Mean Squared Error)** use karta hai. Ye model ki prediction aur asli value ke beech ki doori ka square nikalta hai.
   $$Loss = \frac{1}{n} \sum_{i=1}^{n} (y_{true} - y_{pred})^2$$

#### Parameters
Sabse acchi baat ye hai ki `MLPRegressor` ke baaki saare parameters (`hidden_layer_sizes`, `solver`, `alpha`, `learning_rate`, etc.) **bilkul same** wahi hain jo humne `MLPClassifier` mein padhe the!
