### DictVectorizer
* " from sklearn.feature_extraction import FeatureHasher "
* Convert list of mappings of feature name and feature value , into matrix
* it keep numerical values as it is but for categorical values it uses One-Hot-Encoding
* only list of dict can be passed


```python
data1 = [
    {"age":4 , "height":96.4},
    {"age":1 , "height":73.9},
    {"age":3 , "height":88.9},
    {"age":2 , "height":81.6}
]

data2 = [
    {'city': 'Delhi', 'age': 25},
    {'city': 'Mumbai', 'age': 30},
    {'city': 'Delhi', 'age': 22}
]
```

```python
from sklearn.feature_extraction import DictVectorizer
dv = DictVectorizer(sparse=False)
```

```python
df1 = dv.fit_transform(data1)
df1
```

```python
df2 = dv.fit_transform(data2)
df2
```

### FeatureHasher : High Speed , Low Memory Vectorizer that uses feature hashing technique

* " from sklearn.feature_extraction import FeatureHasher "

* insted of building hash table of features as Vectorizer do , it applies a hash function to the feature to determine their column index in sample matrix directly

* outputs in scipy.sparse matrix
