### Handling Missing Values
* we can use
    * SimpleImputer -> "from sklearn.impute import SimpleImputer"
    * KNNImputer -> "from sklearn.impute import KNNImputer"

#### Simple Imputer
* Can fill up missing values by following Stratigies
    * (1) mean , (2) median , (3) most_frequent , (4) constant

```python
from sklearn.impute import SimpleImputer

X = [
    [7 , 1],
    [None , 8],
    [2 , None],
    [9 , 6]
]

si = SimpleImputer(strategy="mean")
si.fit_transform(X)
```

#### KNN Imputer

```python
from sklearn.impute import KNNImputer

X2 = [
    [2,3,None],
    [3,4,3],
    [None,5,6],
    [8,8,7]
]

knn_i = KNNImputer()
knn_i.fit_transform(X2)
```
