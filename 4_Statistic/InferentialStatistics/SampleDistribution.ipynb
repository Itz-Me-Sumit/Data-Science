```python
import numpy as np
```

```python
class SampleDistribution:

    def __init__(self , population_data , sample , sample_size , population_std = None):

        if len(population_data) == 0:
            raise ValueError("Population Data is Empty")
        else:
            self.data = population_data
        
        if sample==0:
            raise ValueError("You Have to choose that how many samples you want")
        elif sample<0:
            raise ValueError("how sample can be -ve keep it positive")
        else:
            self.sample = sample

        if sample_size==0:
            raise ValueError("You Must have to Choose Sample Size for each Sample")
        elif sample_size<0:
            raise ValueError("how sample_size can be -ve keep it positive")
        else:
            self.sample_size = sample_size

        if population_std and population_std < 0:
            raise ValueError("Standard Deviation Must Be >= 0")
        else:
            self.population_std = population_std

        self.SamplingDistribution()
        

    def SamplingDistribution(self):

        samples = [np.random.choice(self.data , self.sample_size) for _ in range (self.sample)]
        
        self.means = [sample.mean().round(4) for sample in samples]

        return self.means


    def StandardError(self):

        if self.population_std == None:
            one_sample = np.random.choice(self.data , self.sample_size)
            return one_sample.std()/(self.sample_size ** 0.5)
        else:
            return self.population_std/(self.sample_size ** 0.5)
         

```

```python
np.random.seed(42)
data = np.random.normal(loc=165, scale=10, size=500)
```

```python
sd = SampleDistribution(data, sample=100, sample_size=10)
```

```python
print(sd.SamplingDistribution())
```

```python
print(sd.StandardError().round(4))
```

```python
sd = SampleDistribution(data , sample=2000 , sample_size=44)
```

```python
print(sd.SamplingDistribution())
```

```python
print(sd.StandardError().round(5))
```
