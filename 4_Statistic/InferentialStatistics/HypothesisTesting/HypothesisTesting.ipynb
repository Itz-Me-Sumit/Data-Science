```python
import numpy as np


class HypothesisTesting:

    def __init__(
        self,
        sample,
        null_hypothesis,
        alpha=0.05,
        population_std=None,
        tail="two"
    ):

        self.sample = np.array(sample, dtype=float)
        self.null_hypothesis = null_hypothesis
        self.alpha = alpha
        self.population_std = population_std
        self.tail = tail.lower()

        if self.tail not in ("left", "right", "two"):
            raise ValueError("tail must be 'left', 'right' or 'two'")

        self.n = len(self.sample)
        self.sample_mean = self.sample.mean()

        self.test_statistic = self.calculate_test_statistic()


    def standard_error(self):

        if self.population_std is not None:
            return self.population_std / np.sqrt(self.n)

        sample_sd = self.sample.std(ddof=1)
        return sample_sd / np.sqrt(self.n)


    def calculate_test_statistic(self):

        se = self.standard_error()

        stat = (self.sample_mean - self.null_hypothesis) / se

        if self.population_std is not None:
            print(f"Z Statistic : {stat:.4f}")
        else:
            print(f"T Statistic : {stat:.4f}")

        return stat


    def critical_value(self):

        # Z critical values
        if self.population_std is not None:

            z_table = {
                0.10: {
                    "two": 1.645,
                    "left": -1.282,
                    "right": 1.282
                },
                0.05: {
                    "two": 1.960,
                    "left": -1.645,
                    "right": 1.645
                },
                0.01: {
                    "two": 2.576,
                    "left": -2.326,
                    "right": 2.326
                }
            }

            if self.alpha not in z_table:
                raise ValueError("Supported alpha values are 0.10, 0.05, 0.01")

            return z_table[self.alpha][self.tail]

        else:
            raise NotImplementedError(
                "T critical value requires your TDistribution class."
            )


    def decision(self):

        stat = self.test_statistic
        critical = self.critical_value()

        print(f"Critical Value : {critical}")

        if self.tail == "two":

            if abs(stat) > critical:
                print("Decision : Reject Null Hypothesis")
            else:
                print("Decision : Fail to Reject Null Hypothesis")

        elif self.tail == "left":

            if stat < critical:
                print("Decision : Reject Null Hypothesis")
            else:
                print("Decision : Fail to Reject Null Hypothesis")

        else:

            if stat > critical:
                print("Decision : Reject Null Hypothesis")
            else:
                print("Decision : Fail to Reject Null Hypothesis")


    def summary(self):

        print("----------- Hypothesis Test -----------")
        print(f"Sample Size : {self.n}")
        print(f"Sample Mean : {self.sample_mean:.4f}")
        print(f"Null Mean (μ₀) : {self.null_hypothesis}")
        print(f"Alpha : {self.alpha}")
        print(f"Tail             : {self.tail}")

        if self.population_std is None:
            print("Test : One Sample T-Test")
        else:
            print("Test : One Sample Z-Test")

        print(f"Standard Error : {self.standard_error():.4f}")
        print(f"Test Statistic : {self.test_statistic:.4f}")

        self.decision()
```

```python
sample = [52, 49, 51, 48, 50, 49, 51, 50]

test = HypothesisTesting(
    sample=sample,
    null_hypothesis=50,
    population_std=4,
    alpha=0.05,
    tail="two"
)

test.summary()
```
