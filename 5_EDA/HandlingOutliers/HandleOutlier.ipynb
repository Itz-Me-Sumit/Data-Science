### Z-Score

```python
data = [10, 12, 12, 13, 13, 14, 12, 14, 13, 19, 90]
data = pd.DataFrame(data , columns=["Values"])

data["Z-Score"] = data["Values"] - data["Values"].mean() / data["Values"].std()

outliers = data[data["Z-Score"] > data["Z-Score"].mean()]
print(outliers)
```

### Filtering Data on the bases of Inter-Quertile Range

```python
Q1 = data["Values"].quantile(0.25)
Q3 = data["Values"].quantile(0.75)

IQR = Q3-Q1

minimum = Q1 - 1.5 * IQR
maximum = Q3 + 1.5 * IQR

df_iqr = data[ (data["Values"]<minimum) | (data["Values"]>maximum) ]
df_iqr
```
