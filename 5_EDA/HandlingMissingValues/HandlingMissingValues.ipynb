```python
import pandas as pd
import numpy as np

np.random.seed(0)
df = pd.DataFrame({
    "age": np.random.randint(18, 70, size=10).astype(float),
    "salary": np.random.normal(50000, 15000, size=10),
    "city": ["A","B","A","C","B","A","C","B","A","C"],
    "join_date": pd.date_range("2020-01-01", periods=10)
})
# inject some NaNs
df.loc[[1, 2, 5], "age"] = np.nan
df.loc[[2, 7], "salary"] = np.nan
df.loc[[3, 5], "city"] = np.nan

df_backup = df.copy()
df
```

### Simple Imputation

```python
df["age"] = df["age"].fillna(df["age"].median())
```

```python
df["city"] = df["city"].fillna(df["city"].mode()[0])
```

### Handle Time-Series Data

```python
# forward fill -> imputing the value with the previous value
# backward fill -> in backward fill , missing value is imputed using by next value

df_ts = df.sort_values("join_date")
df_ts = df["join_date"].ffill()
df_ts
```
