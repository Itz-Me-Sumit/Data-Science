```python
from imblearn.under_sampling import RandomUnderSampler
from imblearn.over_sampling import RandomOverSampler , SMOTE , ADASYN
from collections import Counter
```

`Sampling_stratagy:-`
* "majority" -> resample the majority class only
* "minority" -> resample the minority class only

```python
# Random Under Sampling

X = df.drop(columns=["readmitted","encounter_id","patient_nbr"] , errors="ignore")
Y = df["readmitted"]

X = pd.get_dummies(X)

rus = RandomUnderSampler(sampling_strategy="not majority" , random_state=42)

x_rus , y_rus = rus.fit_resample(X , Y)
print("Original Distribution",Counter(Y))
print("Under-Sampled Distribution",Counter(y_rus))
```
