```python
categorical_labels = [
    'bedrooms', 'bathrooms', 'stories', 'mainroad','guestroom', 'basement', 'hotwaterheating', 'airconditioning','parking', 'prefarea', 'furnishingstatus'
]
```

### Univatiate Analysis

```python
import math
n_cols = 2
n_rows = math.ceil(len(categorical_labels) / n_cols)

plt.figure(figsize=(12, 4 * n_rows))

for i, label in enumerate(categorical_labels, 1):
    
    plt.subplot(n_rows, n_cols, i)
    sns.countplot(x=label, data=df)
    plt.title(f"{label}")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.boxplot(x=df['area'])

plt.title("Boxplot of Area" , fontsize=14)
plt.xlabel("Area(sq ft)" , fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.kdeplot(df["area"] , fill=True , color="blue" , alpha=0.4 , linewidth=2)

plt.title("KDE Plot of Area" , fontsize=14)
plt.xlabel("Area(sq ft)",fontsize=12)
plt.ylabel("Density" , fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.boxplot(x=df['price'])

plt.title("Boxplot of price" , fontsize=14)
plt.xlabel("Price" , fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.kdeplot(df["price"] , fill=True , color="blue" , alpha=0.4 , linewidth=2)

plt.title("KDE Plot of Price" , fontsize=14)
plt.xlabel("Price",fontsize=12)
plt.ylabel("Density" , fontsize=12)
plt.show()
```

### Bivariate Analysis

```python
numeric_cols = ["area"] # i can add more numerical(continous) columns if avilable

for col in numeric_cols:
    plt.figure(figsize=(8,5))
    sns.scatterplot(x=col , y="price" , data=df , alpha=0.6)
    plt.title(f"Price vs {col}" , fontsize=14)
    plt.xlabel(col , fontsize=12)
    plt.ylabel("price" , fontsize=12)
    plt.show()
```

```python
categorical_cols = [
    'bedrooms', 'bathrooms', 'stories', 'mainroad','guestroom', 'basement', 'hotwaterheating', 
    'airconditioning' ,'parking', 'prefarea', 'furnishingstatus'
]

# Bivariate plots for categorical columns

for col in categorical_cols:
    plt.figure(figsize=(8,3))
    sns.boxplot(x=col , y="price" , data=df)
    plt.title(f"price vs {col}" , fontsize=14)
    plt.xlabel(col , fontsize=12)
    plt.ylabel("price",fontsize=12)
    plt.show()
```

```python
Num_cols = ["price","area","bedrooms","bathrooms","stories","parking"]
plt.figure(figsize=(10,8))
correlation_matrix = df[Num_cols].astype(float).corr()
sns.heatmap(correlation_matrix , annot=True , cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
