### Data Loading

```python
import pandas as pd
data_url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv'
df = pd.read_csv(data_url , sep=";")

features = df.drop("quality" , axis=1)
label = df["quality"].copy()
```

### Data Preperation

```python
from sklearn.model_selection import StratifiedShuffleSplit

# Stratification Of Data
split = StratifiedShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
    )
for train_index , test_index in split.split(features,label):
    strat_train_set = df.loc[train_index]
    strat_test_set = df.loc[test_index]

# Training Dataset
strat_feat_train = strat_train_set.drop("quality",axis=1)
strat_label_train = strat_train_set["quality"].copy()

# Test Dataset
strat_feat_test = strat_test_set.drop("quality",axis=1)
strat_label_test = strat_test_set["quality"].copy()


# Transformation Of Data

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

transform_pipeline = Pipeline([
    ("Imputer" , SimpleImputer(strategy="median")),
    ("std_scaler", StandardScaler())
])

tr_feat_train = transform_pipeline.fit_transform(strat_feat_train)
tr_feat_test = transform_pipeline.transform(strat_feat_test)
```

### Model Training

```python
from sklearn.model_selection import cross_val_score

def DisplayScores(scores):
    print("Scores:",scores)
    print("Mean:",scores.mean())
    print("Standard Deviation",scores.std())
```

### Linear Regression Cross_Validation

```python
# Cross Validation on Linear_Regression Model
from sklearn.linear_model import LinearRegression
lin_reg=LinearRegression()
LinReg_scores = cross_val_score(
    lin_reg,
    tr_feat_train,
    strat_label_train,
    scoring = "neg_mean_squared_error",
    cv=10
)
lin_reg_mse_scores = -(LinReg_scores)
print("Linear Regression Cross Validation:")
DisplayScores(lin_reg_mse_scores)
```

### Decision Tree Regression Cross_Validation

```python
# Cross Validation on Decision_Tree_Regression
from sklearn.tree import DecisionTreeRegressor
tree_reg = DecisionTreeRegressor()
TreeReg_scores = cross_val_score(
    tree_reg,
    tr_feat_train,
    strat_label_train,
    scoring = "neg_mean_squared_error",
    cv=10
)
tree_reg_mse_scores = -(TreeReg_scores)
print("Decision Tree Regression Cross Validation")
DisplayScores(tree_reg_mse_scores)
```

### Random Forest Cross Validation

```python
from sklearn.ensemble import RandomForestRegressor

forest_reg = RandomForestRegressor()

ForestReg_scores = cross_val_score(
    forest_reg,
    tr_feat_train,
    strat_label_train,
    scoring="neg_mean_squared_error",
    cv=10
)

random_forest_mse_scores = -(ForestReg_scores)
print("Random Forest Regressor Cross Validation")
DisplayScores(random_forest_mse_scores)
```
