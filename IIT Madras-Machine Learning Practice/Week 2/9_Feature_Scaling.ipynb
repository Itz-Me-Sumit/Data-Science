```python
from sklearn.model_selection import StratifiedShuffleSplit
import pandas as pd

data_url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv'
df = pd.read_csv(data_url , sep=";")

features = df.drop("quality",axis=1)
label = df["quality"]

split = StratifiedShuffleSplit(n_splits=1 , test_size=0.2 , random_state=42) # Define karta hai ki data ko kaise split karna hai !!

for train_index , test_index in split.split(features,label):
    strat_train_set = df.loc[train_index]
    strat_test_set = df.loc[test_index]

wine_features = strat_train_set.drop("quality" , axis=1)
wine_label = strat_train_set["quality"].copy()
```

### Min-Max Scaling or Normalization
- we subtract minimum value of a feature from the current value and divide it by the difference between the minimum and maximum value of the feature
- bounds values to 0-1

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
result = scaler.fit_transform(wine_features)
print(result)
```

### Standardization
- we substract mean value of each feature from the current value and divide it by the standard daviation so that the resulting feature has a unit variance
- does not bound values to a specific range
- Standardization is less affected by outliers compare to Normalization

```python
from sklearn.preprocessing import StandardScaler

StandardScaler = StandardScaler()
result = StandardScaler.fit_transform(wine_features)
```
