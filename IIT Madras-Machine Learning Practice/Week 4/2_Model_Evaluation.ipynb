### Evaluating Trained Linear Regression Model

After training a Linear Regression model, we need to evaluate:

> Model kitna achha predict kar raha hai.

Scikit-Learn me evaluation ke liye:

```python
linear_regressor.score(x_test, y_test)
```

use karte hain.

---

### score() Method

```python
linear_regressor.score(x_test, y_test)
```

Ye method:

> R² Score (Coefficient of Determination)

return karta hai.

---

### R² Score

R² batata hai:

> Model data ko kitna well explain kar raha hai.

Formula:

```text
                Sum of Squared Errors
R² = 1 -  --------------------------------
          Total Sum of Squared Errors
```

Mathematically:

```text
R² = 1 - [ Σ(yi - ŷi)² / Σ(yi - ȳ)² ]
```

Where:

- `yi` → actual label
- `ŷi` → predicted label
- `ȳ` → mean of predicted label

---

### Residual Sum of Squares (RSS)

Numerator part:

```text
Σ(yi - ŷi)²
```

Ye actual aur predicted values ke beech error batata hai.

- Smaller RSS ⇒ Better model
- Larger RSS ⇒ Poor predictions

---

### Total Sum of Squares (TSS)

Denominator part:

```text
Σ(yi - ȳ)²
```

Ye total variability batata hai dataset me.

---

### R² Score Interpretation

| R² Score | Meaning |
|---|---|
| 1 | Perfect prediction |
| 0 | Mean prediction jitna performance |
| < 0 | Mean se bhi worse |

---

### Example

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()

lr.fit(x_train, y_train)

score = lr.score(x_test, y_test)

print(score)
```

---

### Important Intuition

R² compare karta hai:

```text
Model prediction VS Mean prediction
```

Agar model mean prediction se better perform karta hai, tab R² positive aata hai.

### Common Regression Metrics in `sklearn.metrics`

Scikit-Learn me regression model evaluate karne ke liye bahut saare metrics available hote hain.

Import:

```python
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    root_mean_squared_error,
    r2_score,
    mean_absolute_percentage_error,
    median_absolute_error,
    max_error
)
```

---

### 1. Mean Squared Error (MSE)

```python
from sklearn.metrics import mean_squared_error
```

Formula:

```text
MSE = average of (actual - predicted)²
```

- Large errors ko heavily punish karta hai
- Lower MSE ⇒ Better model

Example:

```python
mse = mean_squared_error(y_test, y_pred)
```

---

### 2. Root Mean Squared Error (RMSE)

```python
from sklearn.metrics import root_mean_squared_error
```

Formula:

```text
RMSE = √MSE
```

- Same unit me error deta hai as target variable
- Easy to interpret

Example:

```python
rmse = root_mean_squared_error(y_test, y_pred)
```

---

### 3. Mean Absolute Error (MAE)

```python
from sklearn.metrics import mean_absolute_error
```

Formula:

```text
MAE = average of |actual - predicted|
```

- Absolute difference leta hai
- Outliers se less sensitive

Example:

```python
mae = mean_absolute_error(y_test, y_pred)
```

---

### 4. R² Score

```python
from sklearn.metrics import r2_score
```

Batata hai:

> Model data ko kitna well explain kar raha hai.

Range:

| Value | Meaning |
|---|---|
| 1 | Perfect prediction |
| 0 | Mean prediction jitna |
| < 0 | Very poor model |

Example:

```python
r2 = r2_score(y_test, y_pred)
```

---

### 5. Mean Absolute Percentage Error (MAPE)

```python
from sklearn.metrics import mean_absolute_percentage_error
```

Formula:

```text
Percentage based error
```

- Error percentage me deta hai
- Business problems me useful

Example:

```python
mape = mean_absolute_percentage_error(y_test, y_pred)
```

---

### 6. Median Absolute Error

```python
from sklearn.metrics import median_absolute_error
```

- Median absolute error calculate karta hai
- Outliers ka impact kam hota hai

Example:

```python
median_error = median_absolute_error(y_test, y_pred)
```

---

### 7. Max Error

```python
from sklearn.metrics import max_error
```

- Maximum single prediction error return karta hai

Example:

```python
mx_error = max_error(y_test, y_pred)
```

---

### Complete Example

```python
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    root_mean_squared_error,
    r2_score,
    mean_absolute_percentage_error,
    median_absolute_error,
    max_error
)

print(mean_squared_error(y_test, y_pred))

print(root_mean_squared_error(y_test, y_pred))

print(mean_absolute_error(y_test, y_pred))

print(r2_score(y_test, y_pred))

print(mean_absolute_percentage_error(y_test, y_pred))

print(median_absolute_error(y_test, y_pred))

print(max_error(y_test, y_pred))
```

---

### Important Intuition

| Metric | Focus |
|---|---|
| MSE | Large errors punish |
| RMSE | Easy interpretation |
| MAE | Absolute average error |
| R² | Overall goodness |
| MAPE | Percentage error |
| Median AE | Robust to outliers |
| Max Error | Worst prediction |

### Cross Validation Techniques in `sklearn.model_selection`

Cross Validation ka use:

> Model ka robust evaluation karne ke liye hota hai.

Instead of:
- sirf ek train-test split

hum:
- multiple splits banate hain
- har split pe model evaluate karte hain

---

### Common Cross Validation Imports

```python
from sklearn.model_selection import (
    KFold,
    StratifiedKFold,
    LeaveOneOut,
    ShuffleSplit,
    RepeatedKFold,
    cross_val_score
)
```

---

### 1. KFold

```python
from sklearn.model_selection import KFold
```

Dataset ko:
- `k` equal parts me divide karta hai.

Example:

```python
kf = KFold(
    n_splits=5
)
```

Process:

```text
Fold 1 -> Test
Remaining -> Train

Fold 2 -> Test
Remaining -> Train
```

and so on...

---

### Example

```python
kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

| Parameter | Meaning |
|---|---|
| `n_splits` | total folds |
| `shuffle` | data shuffle karo |
| `random_state` | reproducibility |

---

### 2. StratifiedKFold

```python
from sklearn.model_selection import StratifiedKFold
```

Mostly:
- classification problems me use hota hai.

Ye ensure karta hai:

> Har fold me class distribution same rahe.

Example:

```python
skf = StratifiedKFold(
    n_splits=5
)
```

---

### 3. LeaveOneOut (LOO)

```python
from sklearn.model_selection import LeaveOneOut
```

### Intuition

```text
1 sample test
```

Very expensive but detailed evaluation.

### Example with KFold

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

kf = KFold(

    shuffle=True,

    y,

```

### Why Cross Validation?

Single train-test split:
- unstable ho sakta hai

Cross Validation:
- multiple evaluations deta hai
- reliable performance batata hai
- overfitting detect karne me help karta hai

````python
# Dataset

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split

x , y = fetch_california_housing(as_frame=True , return_X_y=True)
x_train , x_test , y_train , y_test = train_test_split(x,y , test_size=0.2 , random_state=42)
````

````python
# Cross Validation

from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression

lin_reg = LinearRegression()
cv_score1 = cross_val_score(
    lin_reg,
    x_train,y_train,
    cv=5  # cv parameters uses K-fold cross validation
)

# Alternate way
from sklearn.model_selection import KFold
kfold_cv = KFold(n_splits=5) # we specifically write code for that cv parameter which uses KFold CV
cv_score2 = cross_val_score(
    lin_reg,
    x_train,y_train,
    cv=kfold_cv
)

# We can use another CV method insted of KFold
''' 
from sklearn.model_selection import LeaveOneOut
looCV = LeaveOneOut()
cv_score3 = cross_val_score(
    lin_reg,
    x_train,y_train,
    cv=looCV,
)
'''
````
