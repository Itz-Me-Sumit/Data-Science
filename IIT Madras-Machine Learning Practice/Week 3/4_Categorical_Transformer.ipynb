### OneHotEncoder
* Encodes categorical_features or labels as One-Hot Numeric array
* Creates one-Binary columns for each k unique values
* exactly one column have 1 and rest have 0

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

X = [1,2,3,1]
df = pd.DataFrame(X)
ohe = OneHotEncoder(sparse_output=False)
result = ohe.fit_transform(df)
print(result)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# 1. Employee Dataset
data = {
    'Emp_ID': [101, 102, 103, 104, 105],
    'City': ['Mumbai', 'Delhi', 'Pune', 'Mumbai', 'Delhi'],
    'Department': ['IT', 'HR', 'Finance', 'IT', 'Finance'],
    'Salary': [75000, 60000, 55000, 80000, 65000] # Salary pehle se number hai
}
df = pd.DataFrame(data)

print("--- Original Dataset ---")
print(df)
print("\n")

# 2. Categorical columns (jinme text hai) unhe define karo
categorical_features = ['City', 'Department']

# 3. Professional Approach: ColumnTransformer ka use
# drop='first' se ek column hat jata hai (Dummy Variable Trap se bachne ke liye)
# sparse_output=False hume output array format mein deta hai
preprocessor = ColumnTransformer(
    transformers=[
        ('cat_encoder', OneHotEncoder(drop='first', sparse_output=False), categorical_features)
    ],
    remainder='passthrough' # Baaki columns (Emp_ID, Salary) ko waise hi pass hone do!
)

# 4. Data ko fit aur transform karna
encoded_data = preprocessor.fit_transform(df)

# 5. Naye Columns ke naam dynamically nikalna
encoded_col_names = preprocessor.named_transformers_['cat_encoder'].get_feature_names_out(categorical_features)
# "remainder='passthrough'" ki wajah se jo column bach gaye the, unke naam end me add karenge
all_col_names = list(encoded_col_names) + ['Emp_ID', 'Salary']

# 6. Final Clean DataFrame banana
df_final = pd.DataFrame(encoded_data, columns=all_col_names)

print("--- One-Hot Encoded Dataset ---")
print(df_final)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder

# 1. Ek Realistic Bank Loan Dataset
data = {
    'Applicant_ID': [101, 102, 103, 104, 105],
    'Education': ['High School', 'Masters', 'Graduate', 'High School', 'Masters'], # Feature (X)
    'Loan_Status': ['Rejected', 'Approved', 'Approved', 'Rejected', 'Approved']    # Target (y)
}
df = pd.DataFrame(data)

print("--- Original Dataset ---")
print(df, "\n")

# 2. X (Input) aur y (Output) ko alag karna (Professional Practice)
X = df[['Education']] # 2D array chahiye hota hai features ke liye
y = df['Loan_Status'] # 1D array chahiye hota hai target ke liye

# 3. Features (X) ko Encode karna -> OrdinalEncoder ke sath
# Hum explicitly order bata rahe hain taaki machine galat sequence na seekh le
education_order = [['High School', 'Graduate', 'Masters']] 
ord_encoder = OrdinalEncoder(categories=education_order)
X_encoded = ord_encoder.fit_transform(X)

# 4. Target (y) ko Encode karna -> LabelEncoder ke sath
# Yeh automatically alphabetically encode kar dega (Approved=0, Rejected=1)
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# 5. Final Table banana dekhne ke liye (Real life me hum seedha ML model ko X_encoded, y_encoded de dete hain)
df_final = pd.DataFrame({
    'Applicant_ID': df['Applicant_ID'],
    'Education_Level': X_encoded.flatten(), # .flatten() use kiya taaki 2D wapas 1D ban jaye table ke liye
    'Target_Loan_Status': y_encoded
})

print("--- Machine-Ready Encoded Dataset ---")
print(df_final)
```

### Label Binarizer

* Several regression and binary classification can be extended
to multi-class setup in one-vs-all fashion.

* This involves training a single regressor or classifier per class.

* For this, we need to convert multi-class labels to binary labels,
and LabelBinarizer performs this task.

```python
from sklearn.preprocessing import LabelBinarizer

X = [1,2,6,1,8,6]

lb = LabelBinarizer()

lb.fit_transform(X)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelBinarizer

def process_target_labels():
    # 1. dataset banana (Pandas Series for Target Variable 'y')
    data = {
        'customer_id': [101, 102, 103, 104, 105, 106],
        'subscription_tier': ['Premium', 'Basic', 'VIP', 'Basic', 'Premium', 'VIP']
    }
    df = pd.DataFrame(data)
    
    print("--- Original Target Variable (y) ---")
    print(df[['customer_id', 'subscription_tier']], "\n")
    
    # 2. LabelBinarizer initialize karna
    lb = LabelBinarizer()
    
    # 3. Fit aur Transform karna (Sirf Target column par)
    # Output ek NumPy matrix hota hai
    y_binarized = lb.fit_transform(df['subscription_tier'])
    
    # 4. BEST PRACTICE: Matrix ko wapas DataFrame me badalna taaki columns ke naam dikhein
    # lb.classes_ hume unique sorted labels deta hai (['Basic', 'Premium', 'VIP'])
    y_final_df = pd.DataFrame(y_binarized, columns=lb.classes_)
    
    # Customer ID ke sath final output dikhana
    final_result = pd.concat([df[['customer_id']], y_final_df], axis=1)
    
    print("--- Processed Target Variable with Binarizer ---")
    print(final_result, "\n")
    
    # 5. BONUS: Prediction ke baad wapas original text kaise laayein?
    # Maan lo model ne pehle 2 customers ke liye [0, 1, 0] aur [1, 0, 0] predict kiya
    predictions_from_model = [[0, 1, 0], [1, 0, 0]]
    original_text = lb.inverse_transform(predictions_from_model)
    print("--- Inverse Transform (Matrix back to Text) ---")
    print(f"Matrix {predictions_from_model} \nConverted back to: {original_text}")

process_target_labels()
```
