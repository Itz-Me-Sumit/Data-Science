#### Hyperparameter Tuning Approaches in Scikit-Learn

Scikit-Learn mainly provides two generic methods for Hyperparameter Tuning:

- `GridSearchCV`
- `RandomizedSearchCV`

Both methods use Cross Validation to find the best hyperparameters.

---

#### 1. `GridSearchCV`

`GridSearchCV` checks **all possible parameter combinations**.

Example:

```python
param_grid = [
    {
        'C': [1, 10, 100, 1000],
        'kernel': ['linear']
    }
]
```

Here:

- every value of `C` is tested
- all combinations are evaluated

---

##### How it works?

If parameters are:

```python
C = [1, 10]
kernel = ['linear', 'rbf']
```

then Grid Search tests:

```text
(1, linear)
(1, rbf)
(10, linear)
(10, rbf)
```

---

##### Advantage

- exhaustive search
- better chance of finding optimal parameters

##### Disadvantage

- slow for large parameter spaces

---

#### 2. `RandomizedSearchCV`

`RandomizedSearchCV` randomly samples parameter values from distributions.

Example:

```python
param_dist = {
    "average": [True, False],
    "l1_ratio": stats.uniform(0, 1),
    "alpha": loguniform(1e-4, 1e0),
}
```

---

##### How it works?

Instead of testing all combinations:

- random parameter values are selected
- only limited combinations are tested

---

##### Advantage

- much faster
- works well for large search spaces

##### Disadvantage

- may miss the absolute best parameters

---

#### Key Difference

| GridSearchCV | RandomizedSearchCV |
|---|---|
| Tests all combinations | Tests random combinations |
| Slower | Faster |
| Exhaustive search | Approximate search |
| Better for small spaces | Better for large spaces |

---

#### Final Idea

Both methods try to find the best hyperparameters using Cross Validation.

- `GridSearchCV`
  → exhaustive but slower

- `RandomizedSearchCV`
  → faster and more scalable

#### Difference Between Grid Search and Randomized Search

Both methods are used for Hyperparameter Tuning.

Goal:

---

#### Grid Search

`GridSearchCV` uses exact parameter values.

Example:

```python
param_grid = {
    "alpha": [0.001, 0.01, 0.1]
}
```

It checks **all possible combinations**.

---

##### Characteristics

- exhaustive search
- tries every combination
- more accurate
- slower for large parameter spaces

---

#### Randomized Search

`RandomizedSearchCV` uses parameter distributions.

Example:

```python
param_dist = {
    "alpha": loguniform(1e-4, 1e0)
}
```

Random values are sampled from the distribution.

---

##### Characteristics

- random sampling
- faster
- scalable for large parameter spaces
- computational budget can be controlled

---

#### Computational Budget

In Randomized Search:

```python
n_iter
```

controls:

```text
how many random parameter combinations will be tested
```

Example:

```python
n_iter = 20
```

→ only 20 random combinations are evaluated.

This is independent of:
- number of parameters
- total possible combinations

| Grid Search | Randomized Search |
|---|---|

| Tests all combinations | Tests random combinations |
| Slower | Faster |
| Exhaustive search | Approximate search |
| Better for small search spaces | Better for large search spaces |
