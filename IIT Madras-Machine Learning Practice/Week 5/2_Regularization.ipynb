### Ridge Regularization in Scikit-Learn

Ridge Regression is a type of Regularization used to reduce overfitting in Linear Regression.

It adds a penalty term to the loss function so that model weights do not become excessively large.

Ridge uses **L2 Regularization**:

```text
Loss Function = Original Loss + λ × Σ(w²)
```

Where:

- `λ (lambda)` = regularization strength
- `w` = model weights / coefficients

---

#### Why do we use Ridge Regularization?

Sometimes Linear Regression learns very large coefficients.

This can cause:

- Overfitting
- Poor generalization
- High variance

Ridge Regularization controls coefficient size and makes the model more stable.

---

#### Effect of Regularization

- Small weights → simpler model
- Simpler model → less overfitting
- Better performance on unseen data

---

### Option #1 : Using `Ridge`

This is the direct implementation of Ridge Regression.

#### Code

```python
from sklearn.linear_model import Ridge

ridge = Ridge(alpha=1e-3)
```

---

#### What does `alpha` mean?

`alpha` controls regularization strength.

- Small `alpha`
  → weak regularization
  → behaves similar to normal Linear Regression

- Large `alpha`
  → strong regularization
  → coefficients shrink more

---

#### Important Point

Methods work exactly like normal regression models:

```python
fit()
predict()
score()
```

Example:

```python
ridge.fit(X_train, y_train)
ridge.predict(X_test)
```

---

### Option #2 : Using `SGDRegressor`

Ridge Regularization can also be implemented using SGD.

This is useful for:

- huge datasets
- online learning
- incremental learning

---

#### Code

```python
from sklearn.linear_model import SGDRegressor

sgd = SGDRegressor(
    alpha=1e-3,
    penalty='l2'
)
```

---

#### Important Parameters

##### `alpha`

Controls regularization strength.

##### `penalty='l2'`

Applies Ridge (L2) Regularization.

---

#### Why use `SGDRegressor`?

Unlike normal Ridge Regression, SGD learns iteratively using Gradient Descent.

So it is:

- faster for large datasets
- memory efficient
- suitable for streaming data

---

### Difference Between Both Approaches

| Ridge | SGDRegressor |
|---|---|
| Direct optimization | Gradient Descent optimization |
| Better for small-medium datasets | Better for huge datasets |
| Easy to use | More flexible |
| Stable training | Faster iterative training |

---

### Key Idea

Both methods perform Ridge Regularization.

Difference is only in training approach:

- `Ridge` → direct optimization
- `SGDRegressor` → iterative optimization using Gradient Descent

---

### Final Summary

Ridge Regularization reduces overfitting by shrinking coefficient values.

In Scikit-Learn, it can be implemented using:

```python
Ridge(alpha=...)
```

or

```python
SGDRegressor(alpha=..., penalty='l2')
```

Both apply L2 Regularization but use different optimization techniques.

````python
from sklearn.linear_model import Ridge , SGDRegressor

# Ridge
Ridge = Ridge(alpha=1e-3)

# SGDRegressor
sgd_reg = SGDRegressor(
    alpha=1e-3,
    penalty='l2'
)
````

### How to Search the Best Regularization Parameter for Ridge?

In Ridge Regression, the most important hyperparameter is:

```python
alpha
```

`alpha` controls the strength of regularization.

- Small `alpha`
  → weak regularization
  → model behaves like normal Linear Regression

- Large `alpha`
  → strong regularization
  → coefficients shrink more

Choosing the correct `alpha` is very important because:

- very small `alpha` may cause overfitting
- very large `alpha` may cause underfitting

So we search for the **best regularization parameter** using Cross Validation.

---

### Option #1 : Using `RidgeCV`

`RidgeCV` is a built-in Scikit-Learn estimator that automatically performs Cross Validation to find the best `alpha`.

#### Code

```python
from sklearn.linear_model import RidgeCV

ridge_cv = RidgeCV(
    alphas=[1e-3, 1e-2, 1e-1, 1, 10]
)
```

---

#### What does it do?

It:

- trains Ridge Regression using different alpha values
- performs Cross Validation
- selects the alpha giving best performance

---

#### Advantage

- simple to use
- automatic alpha selection
- less code

---

### Option #2 : Using Cross Validation Manually

We can also search for the best `alpha` manually using:

- Grid Search
- Randomized Search

with:

```python
Ridge
```

or

```python
SGDRegressor
```

---

### Grid Search

Grid Search tries all possible parameter combinations.

Example:

```python
param_grid = {
    "alpha": [1e-3, 1e-2, 1e-1, 1, 10]
}
```

Cross Validation is performed for every alpha value.

Best parameter is selected based on validation performance.

---

#### When to use Grid Search?

Use when:

- parameter space is small
- you want exhaustive search
- higher accuracy is important

---

### Randomized Search

Randomized Search selects random parameter combinations instead of checking all values.

It is faster than Grid Search.

---

#### When to use Randomized Search?

Use when:

- parameter space is very large
- training is expensive
- faster experimentation is needed

---

### Difference Between Both Approaches

| Method | Behavior |
|---|---|
| `RidgeCV` | Built-in automatic cross validation |
| `GridSearchCV` | Exhaustive parameter search |
| `RandomizedSearchCV` | Random parameter search |

---

### Important Idea

Goal of all methods:

```text
Find the alpha value that gives best generalization performance
```

meaning:

- low validation error
- less overfitting
- better prediction on unseen data

---

### Final Summary

Best regularization parameter in Ridge Regression is searched using Cross Validation.

Main approaches:

1. `RidgeCV`
   - automatic and simple

2. `GridSearchCV`
   - exhaustive search

3. `RandomizedSearchCV`
   - faster random search

All methods try to find the optimal `alpha` value for better model performance.

````python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split

x , y = fetch_california_housing(as_frame=True , return_X_y=True)
x_train , x_test , y_train , y_test = train_test_split(x,y , test_size=0.2 , random_state=42)
````

````python
# RidgeCV 
from sklearn.linear_model import RidgeCV
ridge_cv = RidgeCV(
    alphas=[1e-3, 1e-2, 1e-1, 1, 10]
)
ridge_cv.fit(x_train,y_train)
print("Best Alpha:",ridge_cv.alpha_)
````

````python
# GridSearchCV with Ridge
from sklearn.linear_model import Ridge
from sklearn.model_selection import GridSearchCV

ridge=Ridge()

param_grid ={
    "alpha": [1e-3, 1e-2, 1e-1, 1, 10] 
}

grid_search = GridSearchCV(
    estimator=ridge,
    param_grid=param_grid,
    cv=5
)

grid_search.fit(x_train,y_train)

print("Best Params:" , grid_search.best_params_)
````

````python
# RandomizedSearchCV with Ridge

from sklearn.linear_model import Ridge
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import uniform

ridge = Ridge()

param_dist = {
    "alpha": uniform(0.001, 10)
}

random_search = RandomizedSearchCV(
    estimator=ridge,
    param_distributions=param_dist,
    n_iter=20,
    cv=5,
    random_state=42
)

random_search.fit(x_train, y_train)

print("Best Alpha:", random_search.best_params_)
````

````python
# GridSearchCV with SGDRegressor

from sklearn.linear_model import SGDRegressor
from sklearn.model_selection import GridSearchCV

sgd = SGDRegressor(
    penalty='l2',
    random_state=42
)

param_grid = {
    "alpha": [1e-4, 1e-3, 1e-2, 1e-1]
}

grid_search = GridSearchCV(
    estimator=sgd,
    param_grid=param_grid,
    cv=5
)

grid_search.fit(x_train, y_train)

print("Best Alpha:", grid_search.best_params_)
````

#### Perform Ridge Regression With Polynomial Transformation

````python
from sklearn.linear_model import Ridge
from sklearn.preprocessing import PolynomialFeatures , StandardScaler
from sklearn.pipeline import Pipeline

ridge_poly_model = Pipeline([
    ("Polynomial_transform" , PolynomialFeatures()),
    ("scaler", StandardScaler()),
    ("ridge" , Ridge(alpha=1e-3))
])

ridge_poly_model.fit(x_train,y_train)
````

#### Lasso Regularization

Lasso Regression is used to reduce overfitting by shrinking coefficients.

It uses **L1 Regularization**:

```text
Loss = Original Loss + λ × Σ|w|
```

A special property of Lasso:

- some coefficients become exactly `0`
- performs automatic feature selection

---

#### Option #1 : Using `Lasso`

```python

```

- `alpha` controls regularization strength
- larger `alpha` → stronger regularization

Methods work normally:

```python

predict()
score()
```

---

#### Option #2 : Using `SGDRegressor`

```python
from sklearn.linear_model import SGDRegressor

sgd = SGDRegressor(
    alpha=1e-3,
    penalty='l1'
)
```

Here:

- `penalty='l1'`
  → applies Lasso Regularization

- `alpha`
  → controls regularization strength

---

#### Difference

| Lasso | SGDRegressor |
|---|---|
| Direct optimization | Gradient Descent optimization |
| Better for smaller datasets | Better for huge datasets |

---

#### Final Idea

Lasso:
- reduces overfitting
- shrinks coefficients
- removes unimportant features automatically

````python
# Lasso Regression
from sklearn.linear_model import Lasso

lasso = Lasso(alpha=1e-3)
lasso.fit(x_train,y_train)
````

````python
# SGDRegressor with Lasso Regression

from sklearn.linear_model import SGDRegressor

sgd_reg = SGDRegressor(
    alpha=1e-3,
    penalty="l1"
)

sgd_reg.fit(x_train , y_train)
````

#### Lasso Regression in Polynomial Regression

````python
from sklearn.linear_model import Lasso
from sklearn.preprocessing import PolynomialFeatures , StandardScaler
from sklearn.pipeline import Pipeline

lasso_poly_model = Pipeline([
    ("polynomial_transformation", PolynomialFeatures(degree=2)),
    ("std_scaler" , StandardScaler()),
    ("lasso" , Lasso(
        alpha=1e-1,
        max_iter=1000
    ))
])

lasso_poly_model.fit(x_train,y_train)
````

#### `How to perform both lasso and ridge regularization in polynomial regression?`
**Set up a pipeline of polynomial transformation followed by the SGDRegressor with penalty = 'elasticnet'**

````python
from sklearn.linear_model import Lasso
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures

poly_model = Pipeline([
    ('polynomial_transform', PolynomialFeatures(degree=2) ),
    ('elasticnet', SGDRegressor(
            penalty='elasticnet',
            l1_ratio=0.3
            ))
])

poly_model.fit(x_train, y_train)
````
