### Naive Bayes Classifier:

#### 1. Asal Mein Ye Hai Kya? (Core Concept)
Naive Bayes ek classification algorithm hai. Matlab ye data ko categories mein baantne ka kaam karta hai (jaise Spam ya Not Spam). Ye probability ke **Bayes' Theorem** par based hai. 

#### 2. Isko "Naive" (Bholu/Masoom) Kyu Bolte Hain? (The Intuition)
Is algorithm ka naam "Naive" isliye hai kyunki ye ek bohot hi masoom (naive) assumption leta hai jisko hum **Conditional Independence** kehte hain.

Socho hum predict kar rahe hain ki ek PUBG player **"Pro"** hai ya nahi. Yahan "Pro" hamara class variable ($y$) hai. 
Hamare paas features ($x$) hain: 
* $x_1$ = Headshot rate accha hai
* $x_2$ = Movement (Jiggle) fast hai
* $x_3$ = Looting speed fast hai

Real life mein, agar kisi ka headshot rate accha hai, toh chances hain ki uski movement bhi fast hogi (features aapas mein linked/dependent hote hain). Lekin **Naive Bayes assume karta hai ki saare features ek dusre se 100% independent hain.** Usko lagta hai $x_1$ ka $x_2$ se koi lena-dena nahi hai, agar hume pata hai ki result "Pro" ($y$) hai. Ye assumption real life mein sach nahi hota, isiliye isko "Naive" bolte hain!

#### 3. Math Ka Breakdown (Step-by-Step)

Lecture ka wo darawana formula dekho:
$$P(x_i | y, x_1, \dots, x_{i-1}, x_{i+1}, \dots, x_m) = P(x_i | y)$$

Is equation ko parts mein todte hain:

* **Left Side - $P(x_i | y, \text{baki saare features})$:** Ye math mein pooch raha hai, "Probability batao feature $x_i$ (jaise Headshot rate) hone ki, jab hume pata hai ki final class $y$ (Pro) hai **AUR** baki saari details (Movement, Looting speed) kya hain."
* **Right Side - $P(x_i | y)$:** Ye pooch raha hai, "Probability batao feature $x_i$ ki, jab hume **SIRF** class $y$ pata hai."
* **Equal To ($=$):** Ye equation claim kar rahi hai ki dono baatein same hain! Yaani baki features ($x_1, x_2$ etc.) kya kar rahe hain, usse main feature ($x_i$) ki probability par **koi farq nahi padta** agar target ($y$) constant hai. 

Is ek choti si assumption ki wajah se, poore data ko predict karne ka formula bohot aasan ho jata hai. Hame complex matrix calculations nahi karni padti, bas simple probabilities ko multiply karna hota hai:
$$P(X | y) = P(x_1 | y) \times P(x_2 | y) \times \dots \times P(x_m | y)$$
*(Sab features ki probability alag-alag nikalo aur multiply kar do!)*

#### 4. Ye Algorithm Itna Fast Kyu Hai?
Agar algorithm complex hota, toh usko features ke aapas ke relations (covariance/correlation) ko bhi calculate aur store karna padta. Kyunki Naive Bayes aapas ke relation ko puri tarah ignore kar deta hai, isliye CPU ko sirf simple counting aur multiplication karni hoti hai. 

Isliye jab tumhare paas thousands of features hon (jaise Text/NLP mein hazaron words hote hain), toh Naive Bayes sophisticated methods (jaise SVM ya Neural Networks) ke comparison mein extremely fast train hota hai.

### Naive Bayes Classifiers in Scikit-Learn

Jab hum Python mein `sklearn.naive_bayes` module import karte hain, toh wahan hume alag-alag types ke Naive Bayes classifiers milte hain. 

**Basic Concept:** Inn sab classifiers ka main logic (Bayes Theorem aur Independence) ekdum same hai. Inke beech ka *akela difference* yeh hai ki **tumhara data kaisa hai** (kya wo continuous numbers hain, 0/1 hai, ya categories hain). Har data type ke probability nikalne ka math thoda alag hota hai.

Chalo inko ek-ek karke asaan bhasha se mathematical level tak samajhte hain.

#### 1. Gaussian Naive Bayes (GaussianNB)
**Kahan lagana hai:** Jab tumhare features *Continuous Numbers* hon (jaise Height, Weight, Salary, ya Temperature). 
**Basic Intuition:** Ye assume karta hai ki tumhara data ek "Bell Curve" (Normal Distribution) follow kar raha hai. 

**Mathematical Formulation:**
Probability $P(x_i | y)$ nikalne ke liye ye mean ($\mu$) aur variance ($\sigma^2$) ka use karta hai. Equation kuch aisi dikhti hai:
$$P(x_i | y) = \frac{1}{\sqrt{2\pi\sigma^2_y}} \exp\left(-\frac{(x_i - \mu_y)^2}{2\sigma^2_y}\right)$$

*   **$\mu_y$**: Class $y$ ke liye feature $x$ ka mean (average).
*   **$\sigma^2_y$**: Class $y$ ke liye feature $x$ ka variance (data kitna faila hua hai).
*   **Algorithm ka kaam:** Ye bas har class ke liye mean aur variance calculate karke store kar leta hai.

#### 2. Bernoulli Naive Bayes (BernoulliNB)
**Kahan lagana hai:** Jab tumhare features sirf *Binary* hon (0 ya 1, Yes ya No). Jaise NLP mein, text classify karte waqt hum sirf ye check kar rahe hain ki ek particular word document mein hai (1) ya nahi (0).

**Mathematical Formulation:**
Bernoulli distribution ka formula directly data ke hone ya na hone par penalty lagata hai:
$$P(x_i | y) = P(i | y) x_i + (1 - P(i | y)) (1 - x_i)$$

*   Agar $x_i = 1$ (word present hai), toh probability $P(i | y)$ count hoti hai.
*   Agar $x_i = 0$ (word present nahi hai), toh probability $(1 - P(i | y))$ count hoti hai. 

#### 3. Multinomial Naive Bayes (MultinomialNB)
**Kahan lagana hai:** Ye *Text Classification* (jaise Spam Filter, Sentiment Analysis) ke liye industry standard hai. Jab tumhare features "Counts" hon (ek word sentence mein *kitni baar* aaya hai).

**Mathematical Formulation:**
Ye data ko ek multinomial distribution ki tarah treat karta hai. Probability calculate karne ka standard formula:
$$\theta_{yi} = \frac{N_{yi} + \alpha}{N_y + \alpha n}$$

*   **$N_{yi}$**: Class $y$ ke documents mein feature $i$ (word) kitni baar aaya.
*   **$N_y$**: Class $y$ mein total words ka count.
*   **$\alpha$ (Laplace Smoothing)**: Ye bohot important mathematical concept hai. Agar koi naya word test data mein aa jaye jo training mein nahi tha, toh uski probability $0$ ho jayegi (aur poori equation zero ho jayegi). $\alpha$ (usually $\alpha = 1$) add karne se model zero-probability error se bach jata hai.

#### 4. Categorical Naive Bayes (CategoricalNB)
**Kahan lagana hai:** Jab features specifically *Categorical* hon (jaise Colors: Red, Green, Blue; ya Sizes: S, M, L). Agar tum in categories ko number (0, 1, 2) de do, toh Multinomial unhe counts samajh lega jo ki galat hai. Wahan CategoricalNB kaam aata hai.

**Mathematical Formulation:**
Ye mathematically assume karta hai ki har feature ka apna alag categorical distribution hai. Ye Multinomial jaisa hi Laplace smoothing ($\alpha$) use karta hai probabilities nikalne ke liye taki unseen categories par system crash na ho.

#### 5. Complement Naive Bayes (ComplementNB)
**Kahan lagana hai:** Ye MultinomialNB ka ek advanced aur better version hai, especially jab tumhara data **Imbalanced** ho. (Jaise 990 emails Spam hain, par sirf 10 emails Normal hain).

**Basic Intuition & Math:** 
Normal algorithm check karta hai ki "Item ke class $y$ hone ke kitne chance hain". ComplementNB ulti math lagata hai. Ye calculate karta hai ki "Item ke class $y$ **NAHI** hone ke kya chance hain" (isiliye iska naam Complement hai). 

Iska math parameters ko un classes ke basis par calculate karta hai jo target class nahi hain:
$$\theta_{ci} = \frac{\alpha_i + \sum_{j \notin c} x_{ij}}{\alpha + \sum_{j \notin c} N_j}$$
Ye weights ko normalize karta hai jisse imbalanced data ka bias (jyada data wali class ko favor karne ki aadat) khatam ho jata hai. NLP mein ye regularly MultinomialNB ko beat karta hai.
