### Dataset Loader

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

X , y = load_iris(return_X_y = True)
X_train , X_test , y_train , y_test = train_test_split(X , y , test_size = 0.3 , random_state=42)

X_train = torch.from_numpy(X_train).float()
y_train = torch.from_numpy(y_train)
X_test = torch.from_numpy(X_test).float()
y_test = torch.from_numpy(y_test)

from torch.utils.data import TensorDataset , DataLoader

train_dataset = TensorDataset(X_train , y_train)
test_dataset = TensorDataset(X_test , y_test)

BATCH_SIZE = 10
NUM_WORKERS = 2

train_loader = DataLoader(
    dataset = train_dataset,
    batch_size = BATCH_SIZE,
    num_workers = NUM_WORKERS,
    shuffle = True
)

test_loader = DataLoader(
    dataset = test_dataset,
    batch_size = BATCH_SIZE,
    num_workers = NUM_WORKERS,
    shuffle = True
)

for X_batch , y_batch in train_loader:
  print(f"X_train\n{X_batch}\ny_train\n{y_batch}")
```

### Custom Loss Function

```python
import torch.nn as nn

class MeanAbsoluteLoss(nn.Module):
  def __init__(self):
    super.__init__()

  def forward(prediction : torch.Tensor , target : torch.Tensor) -> torch.Tensor:

    abs_error = torch.abs(target - prediction)
    return torch.mean(abs_error)

class huberloss(nn.Module):
  def __init__(self , delta:float = 1.0):
    super.__init__()

  def forward(self , prediction : torch.Tensor , target : torch.Tensor) -> torch.Tensor:
    residuals = prediction - target
    abs_residuals = torch.abs(residuals)

    quadratic_mask = abs_residuals <= self.delta
    linear_mask = abs_residuals > self.delta

    quadratic_loss = 0.5 * residuals**2
    linear_loss = self.delta * abs_residuals - 0.5 * self.delta **2

    loss = quadratic_mask * quadratic_loss + linear_mask * linear_loss

    return torch.mean(loss)
```

### Training Loop

```python
torch.manual_seed(42)


import torch.nn as nn

class LinearRegression(nn.Module):
  def __init__(self):
    super().__init__()
    self.weights = nn.Parameter(
        torch.rand(4 , dtype = torch.float),
        requires_grad = True
    )
    self.bias = nn.Parameter(
        torch.rand(1 , dtype = torch.float),
        requires_grad = True
    )

  def forward(self , x : torch.Tensor) -> torch.Tensor:
    return x @ self.weights +  self.bias



model = LinearRegression()

loss_fn = nn.L1Loss()

optimizer = torch.optim.SGD(
    params = model.parameters(),
    lr = 0.001
)


EPOCHS = 10
PRINT_INTERVAL = 5


# Initializing Traking Variables for Loss curves
train_loss_value = []
test_loss_value = []
epoch_count = []

for epoch in range(EPOCHS):

  model.train()

  epoch_train_loss = 0.0
  num_train_batches = 0

  for X_batch , y_batch in train_loader:

    y_pred = model(X_batch) # Prediction of model

    batch_loss = loss_fn(y_pred , y_batch)

    optimizer.zero_grad()

    batch_loss.backward()

    optimizer.step()

    epoch_train_loss+=batch_loss.item()
    num_train_batches+=1

  avg_train_loss = epoch_train_loss/num_train_batches

  model.eval() # set the model into evaluation mode - it disables gradient computations



  epoch_test_loss = 0.0
  num_test_batch = 0
  with torch.inference_mode():

    for X_batch , y_batch in test_loader:

      y_pred = model(X_batch)

      batch_loss = loss_fn(y_pred , y_batch)

      epoch_test_loss += batch_loss.item()
      num_test_batch += 1

  avg_test_loss = epoch_test_loss/num_test_batch

  if epoch % PRINT_INTERVAL == 0:
    epoch_count.append(epoch)
    train_loss_value.append(avg_train_loss)
    test_loss_value.append(avg_test_loss)

    print(f"""
    Epoch : {epoch} , Avg Train Loss : {avg_train_loss} , Avg Test Loss : {avg_test_loss}
    """)

print("\nTraining Completed")
```

```python
import matplotlib.pyplot as plt

plt.plot(epoch_count , train_loss_value , label = "Train Loss")
plt.plot(epoch_count , test_loss_value , label = "Test Loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Training and Test Loss Curve")
plt.legend()
plt.show()
```
